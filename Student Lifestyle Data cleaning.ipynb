```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("student_lifestyle_performance_dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df = df.dropna()
df.columns = df.columns.str.strip()
```

```python
df['Performance'] = 'Good'
df.loc[df['CGPA']>8.50, 'Performance'] = 'Excellent'
df.loc[df['CGPA']<6.50, 'Performance'] = 'Poor'
```

```python
df.head()
```

## Study Hours VS CGPA


```python
import matplotlib.pyplot as plt
plt.scatter(df['Study_Hours_per_Day'], df['CGPA'])
plt.xlabel = ("Study Hours")
plt.ylabel = ("CGPA")
plt.show()
```

## Sleep VS CGPA

```python
plt.scatter(df['Sleep_Hours'], df['CGPA'])
plt.show()
```

## Attendace VS CGPA

```python
plt.scatter(df['Attendance_Percentage'], df['CGPA'])
plt.show()
```

## Screentime VS CGPA

```python
plt.scatter(df['Screen_Time_Hours'], df['CGPA'])
plt.show()
```

## Internal Marks VS CGPA

```python
plt.scatter(df['Internal_Marks'], df['CGPA'])
plt.show()
```
